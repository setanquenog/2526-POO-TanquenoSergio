 ## Sistemas de Cuentas Bancarias en Python

### ¿De qué se trata el programa?

#### Simula un sistema básico de cuentas bancarias desarrollado mediante el lenguaje Python utilizando la Programación Orientada a Objetos, creando una cuenta bancaria que permite realizar depósitos, retiros y mostrar información de la cuenta, como objetivo de práctica utilizar constructores (_init_) y destructores (_del_), organizando el programa en capas: modelos, servicio y main.

### ¿Cómo ejecutar main.py)

#### Verificar que las carpetas esten correctamente organizada (modelos, servicios y main.py), hacer clic derecho sobre el archivo main.py, seguidamente seleccionar Run "main" y observar la salida del programa en la consola.

### ¿Dónde se evidencia el uso de _init_ y _del_?

#### init (constructor): Se ejecuto cuando se creo la cuenta bancaria, iniciando los datos de la cuenta como número, titular y saldo inicial.

#### del (destructor): Se ejecuta cuando el objeto es eliminado de la memoria, simula el cierre de la cuenta y la liberación de recursos,evidenciándose cuando se utiliza "del" o al terminar la ejecución del programa.
